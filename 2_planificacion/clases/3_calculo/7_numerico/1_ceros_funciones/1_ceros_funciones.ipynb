```python
import pandas as pd
import numpy as np

# Cargar la lista de estudiantes desde el archivo CSV
path = r"C:\Users\marco\Documentos\docencia\groups_list\numerico.xlsx"

estudiantes = pd.read_excel(path) 

estudiantes['Parcial'] = estudiantes['Parcial'].astype(float)  # Asegurar que la columna 'Parcial' es float

# Convertir la columna 'Nombre' a una lista de Python
lista_estudiantes = estudiantes['Nombre'].tolist()

# Función para llamar a un estudiante aleatoriamente y registrar su nota de seguimiento
def asistencia():
    global lista_estudiantes
    if lista_estudiantes:
        estudiante_aleatorio = np.random.choice(lista_estudiantes)
        print(f"Llamando a {estudiante_aleatorio}...")
        asistencia = input(f"¿Asistió {estudiante_aleatorio}? (s/n): ").strip().lower()
        
        # Validación de entrada
        if asistencia == 's':
            # Obtener el índice del estudiante en el DataFrame
            indice = estudiantes[estudiantes['Nombre'] == estudiante_aleatorio].index[0]
            # Incrementar su valor en la columna 'Parcial'
            estudiantes.at[indice, 'Parcial'] += 0.3
            print(f"Actualizado: {estudiante_aleatorio} ahora tiene {estudiantes.at[indice, 'Parcial']} en la columna 'Parcial'.")
        
        elif asistencia != 'n':
            print("Entrada inválida. Por favor, introduce 's' para sí o 'n' para no.")
            return  # Volver a preguntar para el mismo estudiante si la entrada es inválida
        
        # Remover al estudiante de la lista para no llamarlo de nuevo
        lista_estudiantes.remove(estudiante_aleatorio)
    else:
        print("Se han llamado a todos los estudiantes del grupo completo.")

# Función para guardar cambios al final de la sesión
def guardar_cambios():
    estudiantes.to_csv(path, index=False)
    print("Cambios guardados en el archivo CSV.")

# Ejemplo de uso
while lista_estudiantes:
    asistencia()

# Guardar los cambios una vez que todos los estudiantes hayan sido llamados
guardar_cambios()

```

### **1. Introducción a los ceros de funciones y su importancia**  



En matemáticas y ciencias aplicadas, los **ceros de una función** son los valores de la variable independiente para los cuales la función toma un valor de cero. 

Es decir, si $f(x)$ es una función dada, un **cero** de $f(x)$ es cualquier número $x^{*}$ tal que:  

$$
f(x^*) = 0
$$

La determinación de ceros de funciones es un problema fundamental en diversas áreas del conocimiento, ya que permite:  



- **Modelar fenómenos físicos y naturales**: En ingeniería y ciencias aplicadas, los ceros de una función pueden representar puntos de equilibrio, intersecciones de curvas o condiciones de estabilidad.  


- **Resolver ecuaciones no lineales**: Muchas ecuaciones algebraicas y trascendentes no tienen soluciones exactas, por lo que se requiere encontrar aproximaciones numéricas.  


- **Optimización y análisis de sistemas**: En economía, biología, y otras disciplinas, los ceros de ciertas funciones pueden indicar valores críticos donde ocurren cambios significativos en un sistema. 

 
- **Aplicaciones en circuitos eléctricos y mecánica estructural**: La determinación de ceros es clave en el análisis de oscilaciones, circuitos y sistemas dinámicos.  



Dado que muchas funciones no permiten encontrar sus ceros de forma analítica, en esta clase exploraremos **métodos numéricos** como **bisección, Newton-Raphson y falsa posición** para aproximar soluciones con una precisión deseada.

## **1. Introducción a los ceros de funciones y su importancia en Ingeniería Agropecuaria**  



Los **ceros de una función** son los valores de la variable independiente para los cuales la función se anula, es decir, si $f(x)$ es una función dada, un **cero** es un número $x^{*}$ tal que:



$$
f(x^{*}) = 0
$$



En **Ingeniería Agropecuaria**, la determinación de ceros de funciones tiene múltiples aplicaciones, como:  



1. **Hidráulica agrícola**: Para calcular el caudal óptimo en sistemas de riego y resolver ecuaciones de pérdidas de carga. 

 
2. **Crecimiento de cultivos**: Para determinar el momento en el que el crecimiento de una planta se estabiliza o se maximiza.  


3. **Sistemas térmicos**: Para analizar la temperatura óptima del suelo en función del tiempo y encontrar valores donde la variación térmica es cero. 

 
4. **Optimización de fertilización**: Para modelar la respuesta de los cultivos a la fertilización y encontrar el nivel óptimo de nutrientes.  



A continuación, se presentan ejemplos con visualizaciones que ayudan a comprender estos conceptos.

## **Ejemplo 1: Optimización del caudal en un sistema de riego**  



La ecuación de pérdidas de carga en tuberías se puede modelar mediante la ecuación de Darcy-Weisbach:



$$
h_f = f \frac{L}{D} \frac{v^2}{2g}
$$



Donde $h_f$ es la pérdida de carga, $f$ es el factor de fricción, $L$ es la longitud de la tubería, $D$ es el diámetro, $v$ es la velocidad del agua y $g$ es la gravedad.  



Podemos modelar esta ecuación para encontrar la velocidad $v$ óptima donde las pérdidas de carga son aceptables.



### **Visualización con Matplotlib**


```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros del sistema
L = 100  # Longitud de la tubería en metros
D = 0.1  # Diámetro en metros
g = 9.81  # Gravedad m/s^2
f = 0.02  # Factor de fricción

# Rango de velocidades
v = np.linspace(0.1, 5, 100)
hf = f * (L / D) * (v**2 / (2 * g))  # Pérdidas de carga

# Buscar el cero de la función (cuando la pérdida es tolerable, ej. 2m)
hf_target = 2
idx = np.argmin(np.abs(hf - hf_target))

# Gráfica
plt.figure(figsize=(8,5))
plt.plot(v, hf, label='Pérdida de carga $h_f$', color='b')
plt.axhline(hf_target, color='r', linestyle='--', label='Límite aceptable')
plt.scatter(v[idx], hf[idx], color='black', zorder=3, label=f'Velocidad óptima: {v[idx]:.2f} m/s')
plt.xlabel('Velocidad del agua (m/s)')
plt.ylabel('Pérdida de carga (m)')
plt.title('Optimización de velocidad en riego agrícola')
plt.legend()
plt.grid()
plt.show()

```

🔹 **Conclusión**: Encontramos la velocidad óptima donde la pérdida de carga es tolerable.

## **Ejemplo 2: Crecimiento de cultivos y cero de la tasa de crecimiento**  



El crecimiento de los cultivos sigue un modelo logístico:



$$
G(t) = \frac{K}{1 + e^{-r(t - t_0)}}
$$



donde:
- $G(t)$ es la altura del cultivo al cabo de un tiempo $t$. 
- $K$ es el crecimiento máximo del cultivo,
- $r$ es la tasa de crecimiento,
- $t_{0}$ es el tiempo en que la velocidad de crecimiento es máxima.  



Para encontrar el tiempo en que el crecimiento se estabiliza, buscamos el **cero de la derivada** $G'(t)$.



### **Visualización con Seaborn**


```python
import seaborn as sns

# Parámetros del modelo logístico
K = 100  # Crecimiento máximo (en cm)
r = 0.5  # Tasa de crecimiento
t0 = 10  # Momento de crecimiento máximo

# Rango de tiempo
t = np.arange(0, 30, 0.01)
G = K / (1 + np.exp(-r * (t - t0)))  # Crecimiento

# Derivada (tasa de crecimiento)
G_deriv = (r * K * np.exp(-r * (t - t0))) / (1 + np.exp(-r * (t - t0)))**2

# Encontrar el cero de la derivada (cuando G' = 0)
idx_max_growth = np.argmax(G_deriv)

# Gráfico
plt.figure(figsize=(8,5))
sns.lineplot(x=t, y=G, label='Crecimiento del cultivo', color='g')
plt.axvline(t[idx_max_growth], color='r', linestyle='--', \
    label=f'Máximo crecimiento en t={t[idx_max_growth]:.1f}')
plt.xlabel('Tiempo (días)')
plt.ylabel('Altura del cultivo (cm)')
plt.title('Crecimiento de cultivos y tasa de crecimiento')
plt.legend()
plt.grid()
plt.show()
```


🔹 **Conclusión**: Determinamos el momento en que la tasa de crecimiento es máxima, lo que ayuda a decidir cuándo fertilizar o cosechar.

## **Ejemplo 3: Variación térmica del suelo con animación**  

La temperatura del suelo varía con el tiempo y puede modelarse como una onda senoidal:

$$
T(t) = T_m + A \cos(2\pi f t)
$$

donde:
- $T_m$ es la temperatura media,
- $A$ es la amplitud,
- $f$ es la frecuencia.  

Buscamos los momentos del día donde la temperatura cambia más rápido.



### **Animación con Matplotlib**


Aquí tienes el código completo y corregido para que se renderice correctamente en cuadernos Jupyter ejecutados en **VSCode**. Se incluye la importación de **IPython.display.HTML** para asegurar la correcta visualización de la animación.  

### Código completo:


```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

# Parámetros
T_m = 20  # Temperatura media en °C
A = 10    # Amplitud de variación térmica
f = 1/24  # Frecuencia del ciclo diario
t = np.linspace(0, 48, 200)  # Rango de tiempo (2 días)

# Función de temperatura del suelo
T = T_m + A * np.cos(2 * np.pi * f * t)

# Configuración de la figura
fig, ax = plt.subplots(figsize=(8, 5))
line, = ax.plot([], [], 'r', lw=2)
ax.set_xlim(0, 48)
ax.set_ylim(T_m - A - 2, T_m + A + 2)
ax.set_xlabel("Tiempo (horas)")
ax.set_ylabel("Temperatura (°C)")
ax.set_title("Variación térmica del suelo")

# Función de actualización para la animación
def update(frame):
    line.set_data(t[:frame], T[:frame])
    return line,

# Crear animación
ani = animation.FuncAnimation(fig, update, frames=len(t), interval=50, blit=True)

# Mostrar animación en Jupyter Notebook
HTML(ani.to_jshtml())  # Para asegurarse de que se renderice correctamente en VSCode

```


### **Explicación y correcciones**:
1. **Se importó `IPython.display.HTML`** para renderizar la animación en Jupyter Notebooks dentro de VSCode.  
2. **Se usó `ani.to_jshtml()`** en lugar de `plt.show()`, ya que este método es más compatible con la visualización en cuadernos Jupyter.  
3. **Se ajustaron los parámetros de la figura** para mejorar la legibilidad.  
4. **Se corrigió el `update` para evitar errores con la visualización parcial de la curva**.

💡 **Ahora, la animación se visualizará correctamente dentro del cuaderno Jupyter en VSCode. 🚀**

🔹 **Conclusión**: Se observa el cambio térmico del suelo, ayudando a planificar riegos o siembras.
