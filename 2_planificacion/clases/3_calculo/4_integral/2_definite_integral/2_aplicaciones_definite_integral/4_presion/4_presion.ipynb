# Presión y fuerza de un fluido  


Los buceadores saben que mientras más profundo se sumerge un objeto en un fluido, es mayor la presión sobre el objeto. 

La presión se define como la fuerza ejercida por unidad de área en la superficie de un cuerpo. 

Por ejemplo, para una columna de agua que tiene 10 pies de altura y $1\ pulg^{2}$ pesa 4.3 libras, la presión del fluido ejercida a una profundidad de 10 pies de agua es $4.3 libras/pulg^{2}$.* 

A 20 pies, ésta aumentaría a $8.6\ libras/pulg^{2}$ y en general la presión será proporcional a la profundidad a la que esté el objeto en el fluido.

##  DEFINICIÓN DE PRESIÓN DE FLUIDO  

 La presión en un objeto a la profundidad $h$ en un líquido es:  
 
 $$
 P = wh
 $$ 

 donde $w$ es la densidad de peso del líquido por unidad de volumen.

Cuando se calcula la presión del fluido, se puede usar una importante (y sorprendente) ley física llamada el principio de Pascal en honor del matemático francés Blaise Pascal. 


El principio de Pascal establece que la presión ejercida por un fluido a una profundidad $h$ es exactamente igual en todas direcciones. 

Por ejemplo, en la figura 7.68, la presión a la profundidad indicada es la misma para los tres objetos. Como se da la presión del fluido en términos de la fuerza por unidad de área $(P = F/A)$, la fuerza del fluido en una superficie de área $A$ sumergida horizontalmente es 

$$
\text{Fuerza del fluido} = F = PA = (\text{presión})(\text{área})
$$

Como tenemos $\rho = 800$ $kg/m^{3}$. Entonces la densidad de peso del aceite es: 
$$

$$

$h(y) = 1- y$

# 1. Planteamiento físico y elección de coordenadas

* Tomamos el plano de la cara circular (disco) como el plano (x)-(y) con el centro del disco en el origen.
* El eje (y) apunta hacia arriba; por tanto la superficie libre está en (y=R).
* Para un punto del disco con coordenada vertical (y\in[-R,R]), la profundidad medida desde la superficie libre es
  [
  h(y)=R-y.
  ]
* La presión hidrostática en ese punto es
  [
  p(y)=\rho g,h(y)=\rho g (R-y).
  ]

# 2. Elemento de área y formulación de la integral

Consideramos una franja horizontal del disco, a altura (y), de espesor (dy). Su longitud lateral es
[
\text{longitud}(y)=2x=2\sqrt{R^2-y^2},
]
por lo que el área de la franja es
[
dA = 2\sqrt{R^2-y^2}, dy.
]
La fuerza sobre esa franja (debida a presión uniforme sobre la franja) es
[
dF = p(y), dA = \rho g (R-y)\cdot 2\sqrt{R^2-y^2}, dy.
]
La fuerza total sobre la cara circular es la integral
[
F=\int_{-R}^{R} \rho g (R-y),2\sqrt{R^2-y^2}, dy.
]

# 3. Cálculo de la integral (pasos)

Separando términos:
[
F = 2\rho g\left[ R\int_{-R}^{R}\sqrt{R^2-y^2},dy ;-; \int_{-R}^{R} y\sqrt{R^2-y^2},dy\right].
]

* El segundo término es cero porque la integranda (y\sqrt{R^2-y^2}) es función impar y el intervalo es simétrico ([-R,R]).
* El primer término contiene la integral
  [
  \int_{-R}^{R}\sqrt{R^2-y^2},dy = \text{(área de media circunferencia)} = \frac{\pi R^2}{2}.
  ]
  Por tanto
  [
  F = 2\rho g ; R ;\cdot; \frac{\pi R^2}{2} = \rho g \pi R^3.
  ]

# 4. Resultado numérico para (R=1\ \mathrm{m})

Sustituyendo (R=1\ \mathrm{m}), (\rho=800\ \mathrm{kg/m^3}) y (g=9.81\ \mathrm{m/s^2}):
[
F = \rho g \pi R^3 = 800\cdot 9.81 \cdot \pi \approx 24655.219\ \mathrm{N}.
]

Resumen numérico:

* Fuerza sobre cada cara circular: ( \boxed{F \approx 2.4655\times 10^{4}\ \mathrm{N}}.)

# 5. Centro de presión (punto de acción de la fuerza resultante)

Para la posición vertical del punto de aplicación (medida desde el centro, con signo positivo hacia arriba) se obtiene analíticamente
[
y_{cp} = -\frac{R}{4}.
]
Con (R=1) m, (y_{cp}=-0.25\ \mathrm{m}): el centro de presión está 0.25 m por debajo del centro del disco.

# 6. Visualizaciones (código)

He incluido y ejecutado un script en Python/matplotlib que:

* calcula (F) y (y_{cp}) y verifica la integral numéricamente,
* dibuja la **distribución de presión** sobre la cara circular (mapa de presión),
* dibuja la **gráfica (p(y))** a lo largo del diámetro vertical y marca (y_{cp}).

Si quieres ejecutar tú mismo el código, aquí te lo dejo (es el mismo que se ejecutó para generar las figuras):


```python


import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos
R = 1.0                 # radio (m)
rho = 800.0             # densidad (kg/m^3)
g = 9.81                # gravedad (m/s^2)

# Cálculo analítico del resultado
F = rho * g * np.pi * R**3
y_cp = -R/4

print(f"F = {F:.3f} N")
print(f"y_cp = {y_cp:.3f} m (debajo del centro)")

# Malla para la figura
nx = ny = 300
x = np.linspace(-R, R, nx)
y = np.linspace(-R, R, ny)
X, Y = np.meshgrid(x, y)
mask = X**2 + Y**2 <= R**2

H = R - Y               # profundidad medida desde la superficie libre (y=R)
P = np.zeros_like(X)
P[mask] = rho * g * H[mask]

plt.figure(figsize=(6,6))
plt.pcolormesh(X, Y, P, shading='auto')
plt.colorbar(label='Presión (Pa)')
theta = np.linspace(0, 2*np.pi, 400)
plt.plot(R*np.cos(theta), R*np.sin(theta), color='black', linewidth=1)
plt.scatter([0.0], [0.0], marker='o')
plt.scatter([0.0], [y_cp], marker='x', s=80)
plt.arrow(0.0, y_cp, 0.6, 0.0, head_width=0.04, head_length=0.06, length_includes_head=True)
plt.text(0.62, y_cp, 'F', va='center')
plt.gca().set_aspect('equal')
plt.title('Distribución de presión sobre la cara circular')
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.xlim(-1.1*R, 1.1*R)
plt.ylim(-1.1*R, 1.1*R)
plt.show()

# Gráfica p(y)
ys = np.linspace(-R, R, 400)
p_of_y = rho * g * (R - ys)
plt.figure(figsize=(6,4))
plt.plot(p_of_y, ys)
plt.scatter([rho*g*(R - y_cp)], [y_cp], marker='x', s=80)
plt.xlabel('Presión p(y) (Pa)')
plt.ylabel('y (m)')
plt.title('Presión a lo largo del diámetro vertical')
plt.gca().invert_yaxis()
plt.grid(alpha=0.3)
plt.show()

```

# Línea argumentativa (Marco teórico)

El desarrollo del **pensamiento computacional** constituye hoy un eje fundamental en la educación matemática. Como lo plantea **Wing (2006)**, esta forma de razonamiento debe considerarse una competencia transversal y básica, al nivel de la lectura, la escritura y las matemáticas. En consecuencia, fomentar su aprendizaje desde los niveles iniciales hasta la educación superior resulta indispensable para formar profesionales capaces de enfrentar los retos de la sociedad contemporánea.

En el campo de la didáctica de las matemáticas, **Papert (1980, 1996)** había anticipado esta necesidad al concebir las computadoras no como simples herramientas de cálculo, sino como **“objetos para pensar”**, capaces de favorecer la construcción activa de significados matemáticos a través de la programación. Esta visión se alinea con el constructivismo de **Piaget**, que resalta el valor de la experimentación en el aprendizaje, y con la perspectiva socio-constructivista de **Vygotsky**, que otorga un papel central a las herramientas culturales en la mediación del conocimiento.

En este marco, el uso de entornos como **Jupyter Notebooks** y lenguajes como **Python, LaTeX, Markdown e inglés** no debe interpretarse como una rivalidad con la enseñanza tradicional de la matemática, sino como una **ampliación de los registros de representación** que describió **Duval (1993, 2006)**. Según su teoría semiótica, el aprendizaje significativo se potencia cuando el estudiante logra traducir y articular distintos sistemas de representación: gráfico, algebraico, numérico, verbal y, en nuestro tiempo, digital.

De igual manera, el **Enfoque Ontosemiótico de Godino (2002, 2017)** muestra que los significados matemáticos emergen de prácticas diversas, lo que justifica la incorporación de estos lenguajes como recursos que enriquecen las configuraciones de práctica matemática, en sintonía con las exigencias de la investigación y la práctica profesional.

En el ámbito internacional, organizaciones como el **NCTM (2000, 2014)** han subrayado que la tecnología es esencial para la enseñanza y el aprendizaje de la matemática, dado que facilita la experimentación, la validación de conjeturas y la comunicación de ideas. Asimismo, instancias como la **ACM** y el **IEEE** han señalado la importancia de desarrollar habilidades computacionales como requisito fundamental en la formación universitaria, tanto en el sector público como privado.

Por tanto, enseñar matemáticas integrando Python, LaTeX, Markdown e inglés en ambientes como Jupyter Notebooks contribuye a la formación de profesionales con **competencias científicas, comunicativas y tecnológicas**, capaces de comprender la matemática de manera interactiva y experimental, y de responder a las demandas del mundo laboral. Esta integración, lejos de ser inadecuada, constituye un **puente entre la enseñanza académica y las necesidades de la sociedad**, fortaleciendo la pertinencia de los programas universitarios.

Estos enfoques coinciden en otorgar a las herramientas culturales un papel esencial en la mediación del aprendizaje.
