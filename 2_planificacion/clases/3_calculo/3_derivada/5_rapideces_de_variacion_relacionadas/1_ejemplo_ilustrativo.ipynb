# La escalera que se desliza 



Una escalera de mano se apoya contra una pared vertical. 

La base de la escalera comienza a deslizarse alejándose de la pared a una velocidad constante de 2 m/s. 

Si la escalera mide 10 metros de longitud, ¿a qué velocidad se está deslizando la parte superior de la escalera hacia abajo cuando la base de la escalera está a 8 metros de la pared?  



Una escalera de mano se apoya contra una pared vertical. 
La base de la escalera comienza a deslizarse alejándose de la pared a una velocidad constante de 2 m/s. Una escalera de mano se apoya contra una pared vertical. 
La base de la escalera comienza a deslizarse alejándose de la pared a una velocidad constante de 2 m/s. 


Solución: 



Esta es **una animación en Python** que muestra claramente que:



  * **La velocidad horizontal de la base es constante** (2 m/s)


  *  **La velocidad vertical de la parte superior es variable** (depende de la posición)



La animación incluye:



* La escalera deslizándose
* El movimiento horizontal uniforme de la base.
* El movimiento vertical no uniforme de la parte superior. 
* Gráficas en tiempo real de ambas velocidades.

# Modelo matemático (idea clave)



Si:

* Longitud de la escalera: $ L = 10 $
* Base: $ x(t) $
* Altura: $ y(t) $



Entonces:
$$
x^2 + y^2 = 100
$$



Derivando respecto al tiempo:
$$
2x\frac{dx}{dt} + 2y\frac{dy}{dt} = 0
\quad \Rightarrow \quad
\frac{dy}{dt} = -\frac{x}{y}\frac{dx}{dt}
$$


$$\frac{dy}{dt} = -\frac{8(2)}{6}$$

```python
dydt = -8*2/6

dydt 
```


Como $ \dot{x} = 2 ,\text{m/s} $, la velocidad vertical **no es constante**.

# Animación en Python (matplotlib)



```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

L = 10
xp = 2

def escalera(t):
    x = xp * t
    if x >= L:
        return
    y = np.sqrt(L**2 - x**2)

    plt.figure(figsize=(5,5))
    plt.plot([0, x], [y, 0], 'r-', lw=3)
    plt.plot(x, 0, 'bo')
    plt.plot(0, y, 'go')
    plt.xlim(0, 11)
    plt.ylim(0, 11)
    plt.gca().set_aspect('equal')
    plt.grid()
    plt.title("Deslizamiento de la escalera")
    plt.show()

interact(
    escalera,
    t=FloatSlider(min=0, max=4.9, step=0.1, value=1)
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

L = 10
xp = 2
t = np.linspace(0.01, 4.9, 200)

x = xp * t
y = np.sqrt(L**2 - x**2)
yp = -(x / y) * xp

plt.figure(figsize=(7,4))
plt.plot(t, xp*np.ones_like(t), label="x'(t) constante")
plt.plot(t, yp, label="y'(t) variable")
plt.xlabel("Tiempo (s)")
plt.ylabel("Velocidad (m/s)")
plt.title("Velocidades vs tiempo")
plt.grid()
plt.legend()
plt.show()
```


# Qué se observa claramente en la animación



*  **La base se mueve con velocidad constante** (línea azul horizontal).
*  **La velocidad vertical cambia** y aumenta en magnitud cuando la escalera se acerca al suelo.
* - La escalera mantiene siempre su longitud (restricción geométrica).
